```python
#from nr_models import rect
import nr_models as nr
from simulation_core import NR 
import numpy as np
```

```python
n= len(df["Bus ID"])
bus_data = df
```

```python
# Initialize an admittance matrix of size n x n with complex numbers (G + jB)
Y_bus_matrix = np.zeros((n, n), dtype=complex)

# Fill the diagonal and off-diagonal elements
for i in range(n):
    # Self-admittance (diagonal element)
    Y_bus_matrix[i, i] = complex(bus_data.loc[i, 'Gsh (S)'], bus_data.loc[i, 'Bsh (S)'])  # G + jB for self-admittance
    
    # Example for off-diagonal elements (negating the admittance between buses)
    for j in range(n):
        if i != j:
            # Replace this with actual admittance data between bus i and bus j
            # For simplicity, let's assume the off-diagonal element is a simple reciprocal
            admittance_between_buses = 1 / complex(bus_data.loc[i, 'Gsh (S)'], bus_data.loc[i, 'Bsh (S)'])
            Y_bus_matrix[i, j] = -admittance_between_buses  # Negate the admittance for off-diagonal
    
# Convert the Y-bus matrix to a DataFrame for better display
Y_bus_df = pd.DataFrame(Y_bus_matrix, columns=[f'Bus {i+1}' for i in range(n)], index=[f'Bus {i+1}' for i in range(n)])

# Print the Y-bus matrix
print("Y-bus Matrix (Admittance Matrix):")
print(Y_bus_df)
```
